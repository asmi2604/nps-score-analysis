##Importing Libraries

```python
import numpy as np
url = "https://drive.google.com/uc?export=download&id=1c0ClC8SrPwJq5rrkyMKyPn80nyHcFikK"
score = np.loadtxt(url)
print(score)
```

####Q1. What NumPy technique is used to filter scores where respondents are detractors (score≤6)?


```python
score
```

```python
score<=6
```

```python
score[score<=6]
```

```python
score[score<=6].shape[0]
```

```python
promoters = score[score >= 9]
# Number of promoters-
num_promoters = len(promoters)
num_promoters
```

```python
detractors = score[score <= 6]
# Number of detractors-
num_detractors = len(detractors)
num_detractors
```

####Q2. Calculate the percentage of Promoters (where score≥9) from the NumPy array score?

```python
total= score.shape[0]
```

```python
percentage_promoters = (num_promoters/total) * 100
percentage_promoters
```

####Question 3: What does this line of code calculate?                       NPS = percentage_promoters - percentage_detractors

```python
percentage_promoters
```

```python
percentage_detractors = (num_detractors/total) * 100
percentage_detractors
```

```python
NPS = percentage_promoters - percentage_detractors
NPS
```
